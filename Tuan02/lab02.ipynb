```python
#BÀI9
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

# In kết quả sau khi chạy xong 1000 vòng lặp
print("Learned coefficients:")
print(w)
```

```python
#BÀI10
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

# In kết quả sau khi kết thúc 2000 vòng lặp
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
#BÀI11
import torch
import numpy as np

# Yêu cầu 1: Khởi tạo và Reshape Tensor

X = torch.arange(24) # Tạo tensor 24 phần tử từ 0 đến 23
print("X shape ban đầu:", X.shape)

X_4x6 = X.reshape(4, 6)
print("\nReshape 4x6:\n", X_4x6)

X_2x12 = X.reshape(2, 12)
print("\nReshape 2x12:\n", X_2x12)

X_2x3x4 = X.reshape(2, 3, 4)
print("\nReshape 2x3x4:\n", X_2x3x4)


# Yêu cầu 2: Các phép toán cơ bản và thống kê

A = torch.rand(3, 4)
B = torch.rand(3, 4)
# Thêm số nhỏ vào B để tránh lỗi chia cho 0 nếu sinh ngẫu nhiên dính số 0
B = B + 1e-6 

print("\n--- Phép toán ---")
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)

print("\n--- Thống kê theo cột (dim=0) ---")
print("Trung bình theo cột của A:\n", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột của A:\n", torch.std(A, dim=0))


# Yêu cầu 3: Tính Autograd cho đa thức

x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

print("\n--- Autograd ---")
print("Đạo hàm dy/dx (PyTorch):", x.grad.item())
# Tính thủ công: y' = 4x + 5. Tại x = 2 -> 4(2) + 5 = 13.
print("Đạo hàm tính thủ công: 13.0")

# Yêu cầu 4: Huấn luyện tìm đa thức

print("\n--- Tìm hệ số đa thức y = 2x^2 - 3x + 5 ---")
# Tạo dữ liệu thực tế dựa trên hàm y = 2x^2 - 3x + 5
polynomial_new = np.poly1d([2, -3, 5])
N = 100 # Tăng N lên để hội tụ chính xác hơn
X_np = np.random.randn(N, 1) * 5
Y_np = polynomial_new(X_np)

# Tạo ma trận đặc trưng
XX_np = np.hstack([X_np * X_np, X_np, np.ones_like(X_np)])

x_tensor = torch.tensor(XX_np, dtype=torch.float32)
y_tensor = torch.tensor(Y_np, dtype=torch.float32)

# Khởi tạo trọng số w cần học
w = torch.randn(3, 1, requires_grad=True)

# Dùng NAdam optimizer
optimizer = torch.optim.NAdam([w], lr=0.05)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_tensor - y_pred))
    mse.backward()
    optimizer.step()

print("Hệ số thực tế: [2.0, -3.0, 5.0]")
print("Hệ số học được:\n", w.data.numpy().flatten())
```
